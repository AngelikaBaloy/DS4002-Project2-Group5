## IAD Flight Departure Delay Rates: ARIMA & SARIMA



```python
# Packages required for analysis
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
```

```python
# Combined datasets (created from EDA/Cleaning script)
data_24 = pd.read_csv("IAD_2024_Combined.csv")
data_25 = pd.read_csv("IAD_2025_Combined.csv")
```

```python
# Cleaning applied to both years

# This removes cancelled flights. Cancelled flights have a departure time of 0:00 and a delay of 0 minutes.

def clean_flights(data):
    data = data.copy()
    data["Date (MM/DD/YYYY)"] = pd.to_datetime(data["Date (MM/DD/YYYY)"])
    data["Departure delay (Minutes)"] = pd.to_numeric(data["Departure delay (Minutes)"], errors="coerce")
    cancelled = data["Actual departure time"].astype(str).isin(["0:00", "00:00"])
    print("Cancelled flights removed:", cancelled.sum())
    data = data[~cancelled]
    # 1 if the flight left 15+ minutes late (FAA definition of a delay)
    data["Delayed"] = (data["Departure delay (Minutes)"] >= 15).astype(int)
    return data

data_24 = clean_flights(data_24) # Removes 488 cancelled flights
data_25 = clean_flights(data_25) # Removes 222 cancelled flights
```

```python
# Daily series: average departure delay and % of flights delayed

# We'll be looking at daily average departure delay in minutes and the proportion of flights experiencing 15+ minute delays.
# We'll create prediction models for both and compare the accuracies.

def daily_series(data):
    daily = data.groupby("Date (MM/DD/YYYY)").agg(
        avg_delay=("Departure delay (Minutes)", "mean"),
        delay_rate=("Delayed", "mean")
    )
    daily["delay_rate"] = daily["delay_rate"] * 100
    daily.index.name = "Date"
    # Give the index a daily frequency so statsmodels knows the spacing between rows
    daily = daily.sort_index().asfreq("D")
    # Fill any day with no flights from its neighbors
    return daily.interpolate()

daily_24 = daily_series(data_24)
daily_25 = daily_series(data_25)

daily_avg_24 = daily_24["avg_delay"].rename("Average Departure Delay")
daily_rate_24 = daily_24["delay_rate"].rename("Departure Delay Rate")
daily_avg_25 = daily_25["avg_delay"].rename("Average Departure Delay")
daily_rate_25 = daily_25["delay_rate"].rename("Departure Delay Rate")

print(daily_24.head())
```

```python
# Dates to forecast: every day in 2025
dates_2025 = pd.date_range(start="2025-01-01", end="2025-12-31", freq="D")
```

```python
# checking for stationarity in both series

for name, series in [("Average departure delay", daily_avg_24), ("Departure delay rate", daily_rate_24)]:
    result = adfuller(series)
    print(name)
    print('  ADF Statistic: %f' % result[0])
    print('  p-value: %f' % result[1])
# if p > .05, the data are non-stationary

# Both p-values are above .05, so the data are non-stationary.
# This means differencing is required. We will begin by looking at the models' performance on data that is not
# differenced. We will later compare these models to our final SARIMA model fit on the differenced data.
```

```python
# Yearly seasonality: Fourier terms

# Because delays can vary in length throughout the year, a wave may be more appropriate.
def fourier_terms(dates, K=3, period=365.25):
    """Sine/cosine waves of the day of the year.
    K = number of wave pairs: higher K = a more detailed yearly curve."""
    day = dates.dayofyear.values
    terms = {}
    for k in range(1, K + 1):
        terms[f"sin{k}"] = np.sin(2 * np.pi * k * day / period)
        terms[f"cos{k}"] = np.cos(2 * np.pi * k * day / period)
    return pd.DataFrame(terms, index=dates)

exog_24 = fourier_terms(daily_24.index)
exog_25 = fourier_terms(dates_2025)
```

```python
# Model fitting and forecasting

def fit_and_forecast(series, lower=None, upper=None):
    """Fits a baseline ARIMA and a SARIMA model to a 2024 series and forecasts 2025."""

    # Log transform to shrink the influence of extreme days.
    # The shift keeps every value above 0 (some days have negative average delays).
    shift = 1 - series.min()
    series_log = np.log(series + shift)

    # Baseline: plain ARIMA
    arima_fit = ARIMA(series, order=(1, 0, 1)).fit()
    arima_pred = arima_fit.forecast(steps=len(dates_2025))

    # SARIMA: weekly seasonal component (period 7) + yearly Fourier terms
    # We predict the SARIMA will perform better and make more accurate predictions
    sarima_fit = SARIMAX(
        series_log,
        exog=exog_24,
        order=(1, 0, 1),
        seasonal_order=(1, 0, 1, 7),
        trend="c"
    ).fit(disp=False)
    print(sarima_fit.summary())

    forecast = sarima_fit.get_forecast(steps=len(dates_2025), exog=exog_25)
    conf_int = forecast.conf_int(alpha=0.05)

    # Undo the log transform so predictions are back in the original units
    predicted = pd.DataFrame({
        "Date": dates_2025,
        "ARIMA Prediction": arima_pred.values,
        "SARIMA Prediction": np.exp(forecast.predicted_mean.values) - shift,
        "Lower 95%": np.exp(conf_int.iloc[:, 0].values) - shift,
        "Upper 95%": np.exp(conf_int.iloc[:, 1].values) - shift
    })

    # Keep percentages within 0-100
    if lower is not None or upper is not None:
        cols = ["ARIMA Prediction", "SARIMA Prediction", "Lower 95%", "Upper 95%"]
        predicted[cols] = predicted[cols].clip(lower=lower, upper=upper)

    return predicted
```

```python
# Accuracy against the actual 2025 values

def score(predicted, actual, label):
    comparison = predicted.merge(actual.rename("Actual").reset_index(), on="Date")
    # Simple benchmark: predict every 2025 day with the 2024 average
    comparison["2024 Average"] = actual_2024_mean[label]
    rows = []
    for model in ["2024 Average", "ARIMA Prediction", "SARIMA Prediction"]:
        error = comparison["Actual"] - comparison[model]
        rows.append({
            "Model": model,
            "MAE": error.abs().mean(),
            "RMSE": np.sqrt((error ** 2).mean())
        })
    print(f"\n{label}: accuracy on {len(comparison)} days of 2025")
    print(pd.DataFrame(rows).round(2).to_string(index=False))
    accuracy = pd.DataFrame(rows).round(2)
    print(f"\n{label}: accuracy on {len(comparison)} days of 2025")
    print(accuracy.to_string(index=False))
    return accuracy

actual_2024_mean = {
    "Average Departure Delay": daily_avg_24.mean(),
    "Departure Delay Rate": daily_rate_24.mean()
}
```

```python
# Plot predicted vs. actual

def plot_forecast(predicted, actual, title, ylabel, actual_label, filename):
    plt.figure(figsize=(14, 6))

    plt.fill_between(
        predicted["Date"],
        predicted["Lower 95%"],
        predicted["Upper 95%"],
        color="green",
        alpha=0.15,
        label="SARIMA 95% interval"
    ) # fills in 95% confidence interval for SARIMA predictions

    sns.lineplot(
        data=actual.reset_index(),
        x="Date",
        y=actual.name,
        color="blue",
        alpha=0.6,
        label=actual_label
    ) # line for actual 2025 data

    sns.lineplot(
        data=predicted,
        x="Date",
        y="ARIMA Prediction",
        color="gray",
        linestyle="--",
        label="ARIMA (baseline)"
    ) # line for ARIMA predictions

    sns.lineplot(
        data=predicted,
        x="Date",
        y="SARIMA Prediction",
        color="green",
        linewidth=2,
        label="SARIMA Predicted 2025"
    ) # line for SARIMA predictions

    plt.title(title)
    plt.xlabel("Date")
    plt.ylabel(ylabel)
    plt.xticks(rotation=45)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename + ".png", dpi=300, bbox_inches="tight") # saves plot
    plt.show()
```

```python
# Model 1: daily departure delay rate (% of flights delayed 15+ min)

# ARIMA + SARIMA on daily departure delay rate
predicted_rate_2025 = fit_and_forecast(daily_rate_24, lower=0, upper=100)
print(predicted_rate_2025.head())

# compares MAE and RMSE across ARIMA + SARIMA + 2024 average
score(predicted_rate_2025, daily_rate_25, "Departure Delay Rate")

# Visualization of results
plot_forecast(
    predicted_rate_2025,
    daily_rate_25,
    title="Predicted vs. Actual Daily Departure Delay Rates (>15min)",
    ylabel="Flights Delayed 15+ min (%)",
    actual_label="Actual Daily Departure Delay Rates 2025",
    filename="delay_rate_forecast"
)

```

```python
# ADF test after removing yearly seasonality
from statsmodels.regression.linear_model import OLS
import statsmodels.api as sm

adf_deseason_rows = []
for name, series in [("Average departure delay", daily_avg_24), ("Departure delay rate", daily_rate_24)]:
    seasonal_fit = OLS(series.values, sm.add_constant(exog_24.values)).fit()
    deseasonalized = series.values - seasonal_fit.fittedvalues
    result = adfuller(deseasonalized)
    print(name, "(yearly seasonality removed)")
    print('  ADF Statistic: %f' % result[0])
    print('  p-value: %f' % result[1])
    adf_deseason_rows.append({"Series": name, "ADF Statistic": result[0], "p-value": result[1]})

pd.DataFrame(adf_deseason_rows).to_csv("adf_deseasonalized_tests.csv", index=False)

# This cell addresses the non-significant p-values we calculated in a previous cell.
# We will add a SARIMA model to run on this differenced data and predict a trend. We will compare the performance of
# this SARIMA model with the SARIMA that was not fit with differenced data. We predict the SARIMA fit on the differenced
# data will perform the best out of all of our models and will compare the MAE and RMSE.
```

```python
# CODE REVISION: SARIMA with differencing (d = 1)

# ARIMA predictions are unchanged; only SARIMA is refit on differenced data.
# With order=(1, 1, 1), statsmodels differences the log series (today minus yesterday)
# and fits the model to those daily changes, then adds the predicted changes back up
# so the forecasts come out as delay levels.

def fit_sarima_differenced(series, name, lower=None, upper=None):
    """Fits SARIMA with d = 1 to a 2024 series and forecasts 2025."""

    # Same log transform as before
    shift = 1 - series.min()
    series_log = np.log(series + shift)

    sarima_diff_fit = SARIMAX(
        series_log,
        exog=exog_24,
        order=(1, 1, 1),               # d = 1: model the day-to-day change
        seasonal_order=(1, 0, 1, 7),   # weekly seasonality
        trend="n"                      # no constant: with d = 1 it would add a straight-line trend
    ).fit(disp=False)
    print(sarima_diff_fit.summary())

    with open(name + "_sarima_differenced_summary.txt", "w") as f:
        f.write(str(sarima_diff_fit.summary()))

    # Forecasts come back as delay levels (statsmodels adds the predicted changes back up)
    forecast = sarima_diff_fit.get_forecast(steps=len(dates_2025), exog=exog_25)
    conf_int = forecast.conf_int(alpha=0.05)

    # Undo the log transform
    predicted = pd.DataFrame({
        "Date": dates_2025,
        "SARIMA (differenced) Prediction": np.exp(forecast.predicted_mean.values) - shift,
        "Lower 95%": np.exp(conf_int.iloc[:, 0].values) - shift,
        "Upper 95%": np.exp(conf_int.iloc[:, 1].values) - shift
    })

    if lower is not None or upper is not None:
        cols = ["SARIMA (differenced) Prediction", "Lower 95%", "Upper 95%"]
        predicted[cols] = predicted[cols].clip(lower=lower, upper=upper)

    predicted.to_csv(name + "_sarima_differenced_predictions_2025.csv", index=False)
    return predicted

def score_differenced(predicted_original, predicted_diff, actual, baseline, label, name):
    """MAE and RMSE for the 2024 average, ARIMA, SARIMA (d = 0) and SARIMA (d = 1)."""
    comparison = (
        predicted_original[["Date", "ARIMA Prediction", "SARIMA Prediction"]]
        .merge(predicted_diff[["Date", "SARIMA (differenced) Prediction"]], on="Date")
        .merge(actual.rename("Actual").reset_index(), on="Date")
    )
    comparison["2024 Average"] = baseline
    comparison = comparison.rename(columns={"SARIMA Prediction": "SARIMA (d=0) Prediction"})

    rows = []
    for model in ["2024 Average", "ARIMA Prediction", "SARIMA (d=0) Prediction", "SARIMA (differenced) Prediction"]:
        error = comparison["Actual"] - comparison[model]
        rows.append({
            "Model": model,
            "MAE": error.abs().mean(),
            "RMSE": np.sqrt((error ** 2).mean())
        })
    accuracy = pd.DataFrame(rows).round(2)
    print(f"\n{label}: accuracy on {len(comparison)} days of 2025")
    print(accuracy.to_string(index=False))

    accuracy.to_csv(name + "_accuracy_differenced.csv", index=False)
    return accuracy

def plot_forecast_differenced(predicted_original, predicted_diff, actual, title, ylabel, actual_label, filename):
    plt.figure(figsize=(14, 6))

    plt.fill_between(
        predicted_diff["Date"],
        predicted_diff["Lower 95%"],
        predicted_diff["Upper 95%"],
        color="green",
        alpha=0.15,
        label="SARIMA (differenced) 95% interval"
    )

    sns.lineplot(
        data=actual.reset_index(),
        x="Date",
        y=actual.name,
        color="blue",
        alpha=0.6,
        label=actual_label
    )

    sns.lineplot(
        data=predicted_original,
        x="Date",
        y="ARIMA Prediction",
        color="gray",
        linestyle="--",
        label="ARIMA (baseline)"
    )

    sns.lineplot(
        data=predicted_diff,
        x="Date",
        y="SARIMA (differenced) Prediction",
        color="green",
        linewidth=2,
        label="SARIMA (differenced) Predicted 2025"
    )

    # The differenced model's interval widens all year, so limit the y-axis
    # to the range of the actual data to keep the lines readable
    plt.ylim(min(actual.min(), 0) - 5, actual.max() * 1.1)

    plt.title(title)
    plt.xlabel("Date")
    plt.ylabel(ylabel)
    plt.xticks(rotation=45)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename + ".png", dpi=300, bbox_inches="tight")   # save before show
    plt.show()


# Model 1: daily departure delay rate (% of flights delayed 15+ min)

predicted_rate_diff_2025 = fit_sarima_differenced(daily_rate_24, name="delay_rate", lower=0, upper=100)

accuracy_rate_diff = score_differenced(
    predicted_rate_2025,
    predicted_rate_diff_2025,
    daily_rate_25,
    baseline=daily_rate_24.mean(),
    label="Departure Delay Rate",
    name="delay_rate"
)

plot_forecast_differenced(
    predicted_rate_2025,
    predicted_rate_diff_2025,
    daily_rate_25,
    title="Predicted vs. Actual Daily Departure Delay Rates (>15min), Differenced SARIMA",
    ylabel="Flights Delayed 15+ min (%)",
    actual_label="Actual Daily Departure Delay Rates 2025",
    filename="delay_rate_predicted_vs_actual_2025_differenced"
)



# Model 2: average daily departure delay (minutes)

predicted_diff_2025 = fit_sarima_differenced(daily_avg_24, name="avg_delay")

accuracy_avg_diff = score_differenced(
    predicted_2025,
    predicted_diff_2025,
    daily_avg_25,
    baseline=daily_avg_24.mean(),
    label="Average Departure Delay",
    name="avg_delay"
)

plot_forecast_differenced(
    predicted_2025,
    predicted_diff_2025,
    daily_avg_25,
    title="Predicted vs. Actual Average Daily Departure Delay (2025), Differenced SARIMA",
    ylabel="Average Departure Delay (minutes)",
    actual_label="Actual 2025",
    filename="avg_delay_predicted_vs_actual_2025_differenced"
)

# This code repeats the analyses done previously, but the SARIMA is instead fit on the differenced data.
# The output is somewhat similar, though the SARIMA's predictions appear smoother.

# The SARIMA model predicting daily departure delay rates has improved after differencing. The MAE is lower
# than the SARIMA without differencing (d=0). However, the SARIMA model predicting average daily departure delay
# in minutes performed worse after differencing. For this forecasting condition, the SARIMA model performs best
# without differencing, which is an interesting result.

# Our final model is the differenced SARIMA predicting daily departure delay rates. We were short of our 5 MAE goal,
# but differencing this model managed to reduce error.
```
